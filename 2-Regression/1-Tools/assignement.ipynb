```python
import matplotlib.pyplot as plt
from sklearn import datasets, linear_model, model_selection
```

```python
linnerud = datasets.load_linnerud()
print(linnerud.DESCR)

print(linnerud.data.shape)
print(linnerud.feature_names)
print(linnerud.data[0])

print(linnerud.target.shape)
print(linnerud.target_names)
print(linnerud.target[0])
```

```python
print("Regression between waistline and how many situps are accomplished")

Sittups = linnerud.data[:, 1]
Waistline = linnerud.target[:, 1]

X = Sittups.reshape(-1, 1)
y = Waistline

model = linear_model.LinearRegression()
model.fit(X, y)

print("R^2: ", model.score(X, y))
```

```python
plt.scatter(X, y, color='black')
plt.plot(X, model.predict(X), color='blue', linewidth=3)
plt.xlabel('Sittups')
plt.ylabel('Waistline')
plt.title('Regression Analysis')
plt.show()
```
